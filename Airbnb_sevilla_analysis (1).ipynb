## Conclusions

The analysis shows that Airbnb prices in Seville vary depending on neighborhood, room type, capacity, and availability.

* **H1 – Neighborhood:** Average prices differ between neighborhoods. For example, El Porvenir and Triana Oeste show higher average prices than several other neighborhoods with at least 10 listings. This suggests that location is related to Airbnb pricing.

* **H2 – Room type:** Entire homes/apartments have a substantially higher average price than private and shared rooms. Hotel rooms also show a high average price, although there are only 17 listings in this category.

* **H3 – Number of guests:** There is a clear general increase in average price as the accommodation capacity increases. Listings for larger groups tend to have higher prices, although the highest-capacity categories may be affected by a small number of listings and unusually high prices.

* **H4 – Availability:** Prices vary across the different availability levels, but the relationship is not perfectly linear. Listings with high availability have an average price of €163.23, compared with €156.72 for low availability and €131.61 for medium-low availability.

Overall, the results suggest that **location, room type, and accommodation capacity are important factors associated with Airbnb prices in Seville**. Availability also shows differences in average price, although the pattern is less clear.

These findings could help Airbnb hosts and property managers understand how different property characteristics and locations are associated with pricing in the Seville market.
